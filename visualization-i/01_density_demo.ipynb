# Try it: histograms and KDE

A **histogram** chops a numeric variable into bins and counts how many points fall in each bin. A **kernel density estimate (KDE)** smooths those points into a continuous curve. Both answer the same question — *what does this variable look like?* — but the picture depends on choices you make.

- **Bin width** controls how wide each histogram bar is.
- **Bandwidth** controls how much the KDE smooths the data.

Too small, and you mostly see noise. Too large, and real bumps disappear.
