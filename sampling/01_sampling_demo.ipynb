# Try it: sampling distributions

A **population** is the full set of values we care about. A **sample** is the subset we actually observe. From one sample we often compute a summary, such as the sample mean.

If we could draw many random samples of the same size, those sample means would not all be identical. The collection of those means is called the **sampling distribution** of the mean. Larger samples usually make that distribution tighter around the true population mean.
